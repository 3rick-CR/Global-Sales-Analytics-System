# Data Cleaning & Exploratory Analysis - Global Superstore Datasets

## Importar bibliotecas:

```python
from IPython.display import display, Markdown
```

## Leer el archivo:

```python
import pandas as pd

# Cargamos el dataset original 
try:
    ruta = 'https://raw.githubusercontent.com/3rick-CR/Global-Sales-Analytics-System/refs/heads/main/data/raw/Global_Superstore.csv'
    df = pd.read_csv(ruta, encoding='latin1')
    
    print("Se cargo el archivo")
    
except FileNotFoundError:
    print("Error: No encontré el archivo CSV. Revisa que esté dentro de la carpeta 'data'.")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```

## Visualizar filas y columnas:

```python
print(f"Filas detectadas: {df.shape[0]}")
print(f"Columnas detectadas: {df.shape[1]}")
display(df.head())
```

## Normalizar nombres de columna:

```python
#Quita espacios , pasa el texto a minuscula y remplaza valores 
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

print("Nombres de columnas:")
print(df.columns.tolist())
```

## Buscar valores nulos:

```python
# Contamos cuántos nulos hay por columna
nulos = df.isnull().sum()

#visualizamos si existen valores NaN
print("--- Columnas con valores (NaN) ---")
for columna, cantidad in nulos.items():
    if cantidad > 0:
        print(f"La columna '{columna}' tiene {cantidad} valores nulos.")



```

* **Nota:** En esta ocasion el dataset de kaggle viene sin valores nulos, como ejemplo en caso de haberlos tenido podriamos utilizar el metodo df = df.dropna(subset=['Sales', 'Profit']) para eliminar unicamente las filas de las columnas 'Sales', 'Profit' que contengan valores nulos.

## Convertir los datos de columna al tipo de dato correcto:

```python
#Ver el tipo de dato por columna
print(df.dtypes.to_string())
```

```python
#Covertir al tipo de dato correcto 

# Fechas de texto" (object) a "fecha" (datetime) 
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
df['sales'] = df['sales'].astype(float)
```

## Creación de nuevos Indicadores Clave KPIs: 

```python

#Genera una columna que calcula el tiempo que tardo en enviarse el producto
df['days_to_ship'] = (df['ship_date'] - df['order_date']).dt.days

display(Markdown("### Tiempos de Envío (Logística):"))
display(df[['order_date', 'ship_date', 'days_to_ship']].head())
```

```python
#Columna que muestra el valor de los descuentos
df['discount_amount'] = df['sales'] * df['discount']

display(Markdown("### Cálculo de Descuento en Moneda Real:"))
display(df[['sales', 'discount', 'discount_amount']].head())
```

```python
#Columna que muestra cual fue el precio unitario 
df['unit_price'] = df['sales'] / df['quantity']

display(Markdown("### Precio unitario:"))
display(df[['unit_price', 'sales', 'quantity']].head())
```

```python
#Columna de categorizacion o etiquetas segun el desempeño
df['profit_status'] = df['profit'].apply(lambda x: 'Profit' if x > 0 else ('Loss' if x < 0 else 'Break-even'))

display(Markdown("### Validación de Categorización de Desempeño:"))
display(df[['sales', 'profit', 'profit_status']].head(10))
```
